```python
import pandas as pd
import json
import os

BASE = r'd:\Programing\ResolveMaping_final2\data'
NEW_EXCEL = os.path.join(BASE, 'Farmers survey 27-2-2026.xlsx')
OLD_EXCEL = os.path.join(BASE, 'Original_Survey_Data_Complete.xlsx')

print('NEW EXCEL SHEETS:')
xl_new = pd.ExcelFile(NEW_EXCEL)
print(xl_new.sheet_names)
```

```python
new_sheets = {}
for sheet in xl_new.sheet_names:
    df = pd.read_excel(NEW_EXCEL, sheet_name=sheet)
    new_sheets[sheet] = df
    print(f'\n--- Sheet: {repr(sheet)} ---')
    print(f'  Rows: {len(df)}, Cols: {len(df.columns)}')
    print(f'  Columns: {list(df.columns)}')
    print(df.head(3).to_string())
```

```python
print('OLD EXCEL SHEETS:')
xl_old = pd.ExcelFile(OLD_EXCEL)
print(xl_old.sheet_names)
old_sheets = {}
for sheet in xl_old.sheet_names:
    df = pd.read_excel(OLD_EXCEL, sheet_name=sheet)
    old_sheets[sheet] = df
    print(f'\n--- Sheet: {repr(sheet)} ---')
    print(f'  Rows: {len(df)}, Cols: {len(df.columns)}')
    print(f'  Columns: {list(df.columns)}')
```

```python
print('EXISTING ARABIC CSVs:')
arabic_dir = os.path.join(BASE, 'layers', 'Arabic')
for f in sorted(os.listdir(arabic_dir)):
    if f.endswith('.csv'):
        df = pd.read_csv(os.path.join(arabic_dir, f), encoding='utf-8-sig')
        print(f'  {f}: {len(df)} rows, cols={list(df.columns)}')

print('\nEXISTING ENGLISH CSVs:')
english_dir = os.path.join(BASE, 'layers', 'English')
for f in sorted(os.listdir(english_dir)):
    if f.endswith('.csv'):
        df = pd.read_csv(os.path.join(english_dir, f), encoding='utf-8-sig')
        print(f'  {f}: {len(df)} rows, cols={list(df.columns)}')
```

```python
# Deep column comparison: new Excel vs existing Arabic CSVs by theme
theme_map = {
    'Water': 'Water_new.csv',
    'Energy': 'Energy_new.csv', 
    'Food': 'Food_new.csv',
    'General': 'General_Info_new.csv',
    'Regen': 'Regenerative_Agriculture_new.csv',
}
print('\nComparing new Excel sheets vs existing Arabic CSVs...')
for sheet_name, df_new in new_sheets.items():
    # Try to find matching CSV
    for theme, csv_name in theme_map.items():
        csv_path = os.path.join(arabic_dir, csv_name)
        if os.path.exists(csv_path):
            df_old = pd.read_csv(csv_path, encoding='utf-8-sig')
            new_cols = set(df_new.columns)
            old_cols = set(df_old.columns)
            added = new_cols - old_cols
            removed = old_cols - new_cols
            same = new_cols & old_cols
            # Check if sheets match roughly by col overlap
            if len(same) > 3:
                print(f'\nSheet "{sheet_name}" ~ CSV "{csv_name}":')
                print(f'  New rows: {len(df_new)}, Old rows: {len(df_old)}, Diff: {len(df_new)-len(df_old)}')
                if added: print(f'  NEW columns added: {added}')
                if removed: print(f'  OLD columns removed: {removed}')
                if not added and not removed: print(f'  Columns identical ({len(same)} cols)')
```
